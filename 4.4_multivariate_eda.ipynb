```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

pd.set_option('display.max_columns', None) 
```

```python
original_df = pd.read_csv('4.3_dataset_clean_v2.csv')
original_df.head()
```

```python
df = original_df.drop(columns = ['link','address','additional_rooms','floor', 'num_parking','features'])
df.head()
```

```python
df.info()
```

```python
import sweetviz as sv

# Create and generate the report
feature_config = sv.FeatureConfig()
report = sv.analyze(source=df, 
                    target_feat='price',
                    feat_cfg=feature_config)

# Show the report in an HTML file
report.show_html(filepath="sweetviz_report.html",
                 layout='vertical')
```
